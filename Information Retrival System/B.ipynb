# Step 7 Load Synonyms

```python


synonym_file_path ="C:\\Users\\Fatima\\Desktop\\files\\synonyms.txt"
synonyms_dict = {}  # Dictionary to store synonyms

if os.path.exists(synonym_file_path):
    with open(synonym_file_path, 'r') as file:
        lines = file.readlines()
        for line in lines:
            words = line.strip().split(',')
            main_word = words[0]
            synonyms = words[1:]
            synonyms_dict[main_word] = synonyms
else:
    print("Synonym file does not exist!")


    
print("\nSynonyms Dictionary\n")
print(synonyms_dict)


```

# Step 8 Extend User Query

```python

user_query = input("Enter your query: ").split()  #taking user input and spliting it
expanded_query = []
for word in user_query:
    expanded_query.append(word)
    if word in synonyms_dict:
        expanded_query.extend(synonyms_dict[word])

print("Expanded Query")
print(expanded_query)
```
