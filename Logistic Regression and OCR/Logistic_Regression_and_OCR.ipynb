```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])  # X contains input features
y = np.array([0, 0, 0, 1])  #  Y contains corssponding output
```

```python
  def loss_function(y_true, y_pred):
      n_samples = len(y_true)
      epsilon = 1e-15  #used to prevent division by zero
      loss = -(1/n_samples) * (y_true * np.log(y_pred + epsilon) + (1 - y_true) * np.log(1 - y_pred + epsilon)).sum()
      return loss






```

```python
def sigmoid(z):   #Sigmoid function
    return 1 / (1 + np.exp(-z))
```

```python
learning_rate = 0.1   #hyperparameters
iterations = 100
```

```python
weights = np.random.rand(2)  #Random initialization of weights and bias
bias = np.random.rand(1)
```

```python
 m = len(y)  # Number of samples

for i in range(iterations):
    # Calculate the linear combination of features with weights and add bias term
    z = np.dot(X, weights) + bias

    # Apply the sigmoid function to get pred
    y_pred = sigmoid(z)

    # Calculate the loss
    loss = loss_function(y, y_pred)

    # Calculate the gradients of weights and bias
    dw = 1/m * np.dot(X.T, (y_pred - y))
    db = 1/m * np.sum(y_pred - y)

    # Update weights and bias
    weights -= learning_rate * dw
    bias -= learning_rate * db



z= np.dot(X, weights) + bias
pred = sigmoid(z) >= 0.5   # value of threshhold is 0.5 so we can classify between 0 and 1


#first initalize weights and bias randomly,calc linear comb of input featues X and bias and add themin sigmoid func to get pred between 0 and 1.
#then compute gradients of loss with respect to weights and bias, updates weights and bias ,After training  compute the pred for the input features using the
#trained weights and bias.
```

```python
print("WIEGHTS = ", weights)
print("BIAS = ", bias)
print("PREDICTED OUTPUT = ", pred.astype(int))
print("ACTUAL OUTPUT = ", y)
accuracy = np.mean(pred == y)  #accuracy
print("Accuracy = ", accuracy)
```

```python
trainX = np.loadtxt(r'C:\Users\Fatima\Desktop\logisticRegression\trainX.txt')
trainY = np.loadtxt(r'C:\Users\Fatima\Desktop\logisticRegression\trainY.txt')
testX = np.loadtxt(r'C:\Users\Fatima\Desktop\logisticRegression\testX.txt')
testY = np.loadtxt(r'C:\Users\Fatima\Desktop\logisticRegression\testY.txt')

print(trainX.shape)
print(trainY.shape)
print(testX.shape)
print(testY.shape)
```

```python
weights = np.zeros(trainX.shape[1])   #initializing weights and bias
bias = 0
```

```python
learning_rates = [0.1, 0.01, 0.001]
```

```python
# Loop over each learning rate
for eta in learning_rates:

    for i in range(100):
        z = np.dot(trainX, weights) + bias    # linear combination of features with weights and add bias term
        predictions = sigmoid(z)             # add bias to get predictions between 0 and 1
        loss = loss_function(trainY, predictions)   #calc loss

        #Compute gradients....simplified it moree for accuracy
        errors = predictions - trainY # calc difference between predictions and actual labels


        trainX_transposed = trainX.T  # Transpose the training data matrix
        gradient_weights = np.dot(trainX_transposed, errors)   #dot product of transposed trainX and errors

        dw = gradient_weights / len(trainY) #gradients for wieghts
        db = np.mean(errors)  # Gradient for bias

        weights -= eta * dw  # Update weights + bias  with eta and gradients
        bias -= eta * db



    # Evaluate the model on the training data
    train_predictions = sigmoid(np.dot(trainX, weights) + bias)
    train_accuracy = np.mean((train_predictions >= 0.5) == trainY) * 100
    print(' Training Accuracy with learning rate =', eta, ':', train_accuracy)

    # Evaluate the model on the testing data
    test_predictions = sigmoid(np.dot(testX, weights) + bias)
    test_accuracy = np.mean((test_predictions >= 0.5) == testY) * 100
    print(' Testing Accuracy with learning rate =', eta, ':', test_accuracy)

```
