```python
 import numpy as np
import matplotlib.pyplot as plt
```

```python
def random_centroids(num_clusters):
    return np.random.rand(num_clusters, 3) * 255
#this function will return an array where each row corresponds to a
#centroid with RGB color values.these centroids are randomly generated
```

```python
def assign_pixels_to_nearest_centroid(pixels, centroids): #cals Euclidean distance between each pixel and centroids
    dist = np.linalg.norm(centroids[:, np.newaxis, :] - pixels, axis=2)
    clusters = np.argmin(dist, axis=0) # Assign each pixel to the nearest centroid based on distance

    return clusters
```

```python
def update_centroids(pixels, assignment, clusters):
    new_centroids = np.zeros((clusters, 3))
    for i in range(clusters):
        cluster_pixels = pixels[assignment == i]
        if len(cluster_pixels) > 0:
            new_centroids[i] = np.mean(cluster_pixels, axis=0)
    return new_centroids
#iterates through each cluster and calculate the mean of pixels assigned to that cluster
#and updates the centroid accordingly
```

```python
 def k_means_clustering(image_name, num_clusters):

    image = plt.imread(image_name) #load img

    print("Original Image Shape:", image.shape)

    pixel_values = image.reshape(-1, 3)  #convert into 2d
    print("Converted Image Shape:", pixel_values.shape)

    centroids = random_centroids(num_clusters)  #initalize clusters randomly

    #Kmeans clustering
    max_iter = 100
    for i in range(max_iter):
        old_centroids = centroids.copy()
        cluster_assignments = assign_pixels_to_nearest_centroid(pixel_values, centroids) #assigns each pixel to the nearest centroid
        centroids = update_centroids(pixel_values, cluster_assignments, num_clusters) #updates the centroids on basis of pixels
        if np.allclose(old_centroids, centroids):
            break

    #here segmented_image will have the same shape as the original image but with pixel values replaced by the centroids
    segmented_image = centroids[cluster_assignments].reshape(image.shape)

   #Display
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(image)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(segmented_image.astype(np.uint8))
    plt.title("Segmented Image (k={})".format(num_clusters))
    plt.axis("off")

    plt.show()



# Kmeans clustering stpes:
#load image : plt.imread(image_name)
#reshape image :we are getting pixel value by converting 3d into 2d :pixel_values = image.reshape(-1, 3)
#random initalization of centroids :random_centroids(num_clusters)
#assign pixels to nearest centroid : cluster_assignments = assign_points_to_nearest_centroid(pixel_values, centroids
#updates centroids on basis of pixels : update_centroids(pixel_values, cluster_assignments, num_clusters)
#segmented image.....image will have the same shape as the original image but with pixel values replaced by the centroids
```

```python
k_means_clustering( 'many.jpg', 3)
k_means_clustering( 'many.jpg', 10)
k_means_clustering( 'many.jpg', 20)
```
