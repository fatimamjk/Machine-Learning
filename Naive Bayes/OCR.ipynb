### Finding the Probabilities

### Training Data

##### Probabailty of 1s and 0s

```python
# as prob is find by divind the the count of its occurences by total number so,
image2_sum = img2_X_train.shape[0]; #total of image 2 and image 4
image4_sum = img4_X_train.shape[0];

image2_one_occ = np.sum(img2_X_train == 1, axis=0);  #occurences of 0s and 1s in image 1 and 2 is calculated
image2_zero_occ = np.sum(img2_X_train == 0, axis=0);
image4_one_occ = np.sum(img4_X_train == 1, axis=0);
image4_zero_occ = np.sum(img4_X_train == 0, axis=0);

#probabilities
img2_one_prob = image2_one_occ / image2_sum;
img2_zero_prob = image2_zero_occ / image2_sum;
img4_one_prob = image4_one_occ /  image4_sum;
img4_zero_prob = image4_zero_occ / image4_sum;

print(img2_one_prob.shape)
print(img2_zero_prob.shape)
print(img4_one_prob.shape)
print(img4_zero_prob.shape)
```

Expected Output <br>
- (256,)
- (256,)
- (256,)
- (256,)


#### Class Probabilities

```python
# Your code here
#Numpy function np.bincount() will be used here. it calculates the occurrence of integers in an input array and
# creates an array where index represents that integer and the value represents the occurences of that integer in an input array.

sum_of_elements= len(Y_train); #will return total no. of elements in Y_train

occ_of_elements = np.bincount(Y_train); #occurrence of elemets is counted


prob_class_2 = occ_of_elements [2] / sum_of_elements;
prob_class_4 = occ_of_elements [4] / sum_of_elements;


print("Probability of class 2:", prob_class_2)
print("Probability of class 4:", prob_class_4)
```

#### Predictions

```python
  # your code here

predicted_array = np.zeros(len(X_train));

#Iterate through each image,
#iterate through each pixel value of the image and cal conditional prob then update predicted classes
for i in range(len(X_train)): #image iteration
    prob_class_2 = occ_of_elements [2] / sum_of_elements;
    prob_class_4 = occ_of_elements [4] / sum_of_elements;

    for j in range(len(X_train[i])):
        if X_train[i][j] == 1:
            prob_class_2 *= img2_one_prob[j]
            prob_class_4 *= img4_one_prob[j]
        else:
            prob_class_2 *= img2_zero_prob[j]
            prob_class_4 *= img4_zero_prob[j]


    predicted_array[i] = 2 if prob_class_2 > prob_class_4 else 4



print("Predicted Classes:", predicted_array)


```

#### Overall Accuracy

```python
# Your code here
#TP =true pos, FP= false pos, FN= false neg,TN=true Neg
TP = 0;
FP = 0;
FN = 0;
TN = 0;

#zip(Y_train, predicted_array) iterates through the pairs of trueValues(X_train) and predicted Values(predicted_array)
# it will pair each element of X_train with each element of predicted_array
for trueValues, predictedValues in zip(Y_train, predicted_array):
    if trueValues == 2 and predictedValues == 2:    #correct prediction
        TP += 1
    elif trueValues == 4 and predictedValues == 2:
        FP += 1
    elif trueValues == 2 and predictedValues == 4:
        FN += 1
    elif trueValues == 4 and predictedValues == 4:   #correct prediction
        TN += 1

#overall accuracy
overall_Accuracy = (TP + TN) / len(Y_train)

#confusion matrix
print("Confusion Matrix");
print("TP =", TP, "FP =", FP);
print("FN =", FN, "TN =", TN);




print("\nOverall Accuracy = ", overall_Accuracy)
```

Solution :

Naive Bayes  works well for OCR tasks as they are  computationally efficient ,making it suiltable for large scale applications where processing speed is crucial.Secondly, it also works well as it assumes that features are conditionllly indpenedent.Naive bayes also allows easy integration of features such as pixel values and it relitively simple to implement
For weather forecasting tasks, machine learning models that can handle complex relationships, temporal dependencies, non-linearities, and diverse data types are often more suitable.Hence, Naive bayes can not be used for weather forecasting tasks.
